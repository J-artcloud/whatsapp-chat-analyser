```python

#parsing the uploaded chat 
import re
import pandas as pd 

chat_dataset = []

with open("data/chat_01.txt", "r", encoding="utf-8") as chat:

    pattern = r"(\d{1,2}/\d{1,2}/\d{2}),\s*(\d{1,2}:\d{2}\s*[AP]M)\s*-\s*(.*?):\s*(.*)"

    file = chat.read()
    for line in file.splitlines():
        match = re.match(pattern, line)

        if match:
            chat_dataset.append({
            "date": match.group(1),
            "time": match.group(2),
            "sender": match.group(3),
            "message": match.group(4)
            })

        elif chat_dataset:
            chat_dataset[-1]["message"] += " " + line.strip()

df = pd.DataFrame(chat_dataset)
df.head()
```

```python
#converting dates and time in strings 
from datetime import datetime 

df["timestamp"] = pd.to_datetime(df["date"] + " " + df["time"], format="%m/%d/%y %I:%M %p")
df['timestamp']
```

```python
# determining the chat time spand
df['timestamp'].diff()
```

```python
df['sender'].value_counts()
df[df["message"].str.contains("omitted", case=False)]['message'].unique()
```

```python
#building a new table to track media presence 
df['has_media'] = df['message'] == '<Media omitted>'
df['has_media'].sum()
```

```python
# what time is the chat most active 
df['timestamp'].dt.hour.value_counts().sort_index()
```

```python
# determinig the number of conversation within each day
df['conv_id'] = (df['timestamp'].diff() > pd.Timedelta(hours= 3)).cumsum()
df[df['conv_id'] == 1]['message']
df['conv_id'].nunique()
df.groupby('conv_id').size()
```

```python
df['word_count'] = df["message"].str.split().str.len()
df["word_count"].describe()
```
